## Part 1: Highest two digits

We have a simple set of lines, each of which is independent of another:

```python
with open('03.txt') as f:
    lines = [line.strip() for line in f.readlines()]
```

For each line, the 'maximum joltage' is the largest possible two-digit number formed: it must be comprised of two digits in the order from the line.

It stands to logic that rather than a naive quadratic search, we can find the highest digit (not including the final), and then the highest digit that remains:

```python
def max_digit(string: str):
    digit = max(string)
    return digit, string.index(digit)

def two_joltage(line: str):
    digit1, index1 = max_digit(line[:-1])
    digit2 = max(line[index1+1:])
    return int(digit1 + digit2)
```

```python
sum(map(two_joltage, lines))
```

## Part 2: Highest twelve digits

The new requirement is to find the highest _twelve_ digits in sequence.

As the first digit is most important for the highest number, it stands to reason we can iteratively search for it; so long as we remember how many digits must remain in the string to not search to the very end.

```python
def twelve_joltage(line: str):
    digits: str = ''
    index = 0
    for d in range(12):
        max_index = len(line) - (12-d) + 1
        digit, offset = max_digit(line[index:max_index])
        digits += digit
        index += offset + 1
    return int(digits)
```

```python
sum(map(twelve_joltage, lines))
```

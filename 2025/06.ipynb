## Part 1: Transposing and applying an operation

A table is input with vertical column of integers ending with either `*` or `+`. We therefore simply transpose (`zip(*iterable)`) and `reduce` the operator, then sum up over every operation:

```python
from operator import mul, add
from functools import reduce
from typing import Iterable

def transpose[T](iterable: Iterable[Iterable[T]]) -> Iterable[Iterable[T]]:
    return zip(*iterable)
```

```python
with open('06.txt') as f:
    lines = f.readlines()
```

```python
sum(
    reduce(
        (mul if op == '*' else add),
        (int(c) for c in ints))
    for *ints, op in transpose(l.split() for l in lines)
) == 5171061464548
```

## Part 2: Transposing in the integers
Now we transpose in the integer space. Notably, this involves precise columnar alignment, so for the first time we'll have to re-parse our input.

As the operator is always on the left of the column, we can "split" by operator to get column lengths:

```python
slices = []
i = 0
final_line = (lines[-1] + ' ').replace('*', '+').split('+')[1:]
for x in final_line:
    slices.append(slice(i, i+len(x)))
    i += len(x) + 1

sum(
    reduce(
        mul if op.strip() == '*' else add,
        (int(''.join(c)) for c in transpose(ints))
    ) for *ints, op in  transpose([l[sl] for sl in slices] for l in lines)
) == 10189959087258
```

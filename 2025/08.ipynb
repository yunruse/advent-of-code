## Part 1: Naive distances

We will connect a circuit based on the smallest distance pair. There's no way about this beyond generating all pairs of coords and getting their distances:

```python
with open("08.txt") as f:
    boxes = [tuple(map(int, line.split(','))) for line in f.readlines()]
```

```python
def pairs[T](iter: list[T]):
    for i, a in enumerate(iter):
        for b in iter[i+1:]:
            yield a, b

def distance(a: tuple[int, ...], b: tuple[int, ...]) -> float:
    return sum((ax-bx)**2 for ax, bx in zip(a, b)) ** 0.5

distances = sorted(
    (distance(*pair), pair) for pair in pairs(boxes)
)
```

We will now connect coordinates by greedily looking at the shortest-distance pairs and connecting them to "circuits", though these circuits do not necessitate being a circular chain.

We'll use a very naive method to fetch this:

```python
circuits = []

for _, (a, b) in distances[:1000]:
    circuit = {a, b}

    for i, c in enumerate(circuits):
        if a in c or b in c:
            circuits.remove(c)
            circuit |= c
    
    circuits.append(circuit)

for sa, sb in pairs(circuits):
    assert sa.isdisjoint(sb), (sa, sb)

s1, s2, s3 = sorted(map(len, circuits))[-3:]
print(s1 * s2 * s3)
```

## Part 2: Keep connecting
For this part, we connect until all circuits are connected. I don't know why this needs to work differently... this one was slightly confusing for me also.

```python
circuits = []

for _, (a, b) in distances:
    circuit = {a, b}
    a_container = set()
    b_container = set()
    for i, c in enumerate(circuits):
        if a in c:
            a_container = c
        if b in c:
            b_container = c
    
    if a_container:
        circuits.remove(a_container)
    if b_container and b_container != a_container:
        circuits.remove(b_container)
    circuits.append(circuit | a_container | b_container)

    if len(circuits[0]) == len(boxes):
        break

a[0] * b[0]
```

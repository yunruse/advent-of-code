## Part 1: Inclusive ranges

Our input comprises two datasets: a set of inclusive integer ranges and a set of numbers:

```python
with open("05.txt") as f:
    range_txt, check_txt = f.read().split('\n\n')
    range_ints = [tuple(map(int, line.split('-'))) for line in range_txt.split('\n')]
    ranges = [range(a, b+1) for a, b in range_ints]
    checks = [int(x) for x in check_txt.split('\n')]
```

A number is "fresh" if it is in one of the ranges:

```python
def is_fresh(x: int):
    return any(x in r for r in ranges)

sum(map(is_fresh, checks))
```

## Part 2: Size of adjoints
Our new answer is to consider every possible integer. The best approach is to add each range in turn, "shrinking" it to avoid overlaps (or skipping it if it is fully contained in another range):

```python
rmods = []
for i, r_new in enumerate(ranges):
    # we will use <> to represent r_new and {} to represent r_check
    for j, r_check in enumerate(rmods):
        if r_check.start <= r_new.start and r_new.stop <= r_check.stop:
            # { <> }
            break

        elif r_check.start <= r_new.start <= r_check.stop:
            # { < } >
            r_new = range(r_check.stop, r_new.stop)

        elif r_check.start <= r_new.stop <= r_check.stop:
            # < { > }
            r_new = range(r_new.start, r_check.start)
        
        elif r_new.start <= r_check.start and r_check.stop <= r_new.stop:
            # < {} >
            del rmods[j]
    else:
        if r_new.start < r_new.stop:
            rmods.append(r_new)

print(sum(map(len, rmods)))
```

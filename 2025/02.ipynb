## Part 1: Repeated IDs

We are to check over a set of ranges separated by commas; a range is two integers separated by a dash, which is inclusive

```python
from itertools import chain
with open("02.txt") as f:
    lims = [map(int, series.split('-')) for series in f.read().strip().split(',')]
    ranges = list(chain(*[range(a, b+1) for a, b in lims]))
len(ranges)
```

An "invalid ID" is one where it has repeating digits, such as `1010`. There are no leading zeroes: `101` is perfectly valid.

Converting back to strings would be painfully slow, so in numerical terms, if `d` digits repeat, then a repeating number `x` holds the following:
- `x` has no leading digits: is greater than `10**(d*2-1) + 10**(d-1)` (11, 1010, 100100)
- `x` repeats: is 0, modulo `10**d + 1` (11, 101, 1001)
- `x` is in that bound: is less than `10**(d*2)` (100, 10000, 1000000)

```python
checks = [(
    10**(d*2-1) + 10**(d-1),
    10**d+1,
    10**(d*2)
    ) for d in range(1, 20)]
checks[:3]
```

Our answer is the sum of all invalid IDs:

```python
sum = 0
for num in ranges:
    for min, mod, max in checks:
        if num < min:
            break # no need to check higher digit lengths
        if num < max and num % mod == 0:
            sum += num
sum
```

## Part 2: Multiply repeating

Now we check repetitions beyond just 2. At this point, rather than do odd O(m^2) checks over all possible digit lengths and repetitions, we can invoke a tiny bit of inefficiency and simply cast to string. For a string of length `lxs` we want all divisors:

```python
sum = 0
for x in ranges:
    xs = str(x)
    lxs = len(xs)

    for d in range(1, lxs//2 + 1):
        # ensure _d_ divides _lxs_
        reps, leftover = divmod(lxs, d)
        if leftover:
            continue
        if xs == xs[:d] * reps:
            sum += x
            break # so we don't trigger eg 1111 in two ways

sum
```

## Part 1

```python
with open("12.txt") as f:
    *shape_txt, region_txt = f.read().split('\n\n')

Shape = list[list[bool]]

shapes: list[Shape] = [
    [
        [c == '#' for c in l]
        for l in s.split('\n')[1:]]
    for s in shape_txt
]

Region = tuple[tuple[int, int], list[int]]
regions: list[Region] = []
for r in region_txt.split('\n'):
    size, shapes_included = r.split(': ')
    w, h = size.split('x')
    shapes_included = list(map(int, shapes_included.split()))
    regions.append(
        ((int(w), int(h)), shapes_included))

def draw(shape: Shape):
    return '\n'.join(
        "".join('#' if b else '.' for b in l)
        for l in shape
    )
```

A good heuristic we can't ignore is that of area:

```python
shape_areas = [
    sum(sum(l) for l in s)
    for s in shapes
]

def matches_area_heuristic(region: Region):
    (w, h), shapes_used = region
    area_given = w * h
    area_needed = sum(
        A * N
        for A, N in zip(shape_areas, shapes_used)
    )
    return area_given >= area_needed

sum(map(matches_area_heuristic, regions))
```

It feels dirty, but this actually works.

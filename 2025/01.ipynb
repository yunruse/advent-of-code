## Part 1: Rotations

The input contains a set of lines which contain `L` or `R` followed by an integer. As left reduces the number, this is just a list of integers:

```python
with open('01.txt') as f:
    turns = [
        int(line[1:]) * (-1 if line[0] == 'L' else +1)
        for line in f.readlines()
    ]
```

A dial starts at 50, and rotates modulo 100. The 'password' is the number of times a rotation leaves us at zero:

```python
N = 0
x = 50
for turn in turns:
    x += turn
    if x % 100 == 0:
        N += 1
N
```

## Part 2

The new pasword is the number of times a rotation _passes_ zero, whether at the end of a rotation or during one.

Recall that our `x` was not bounded but changes continuously, passing into different "zones" where `x // 100`, our "rotation number", changes. In other words, we can take the change in that and call it `r`: its absolute value is the number of times we pass zero.

Each turn we add `r` _in addition_ to 1 if we land on a zero.

However, as a 'zero' is at the start of its 'zone', there are two situations in which a zero is "double counted":
- If we start at zero and move down (eg, 100 to 95)
- If we end at a zero going upwards (eg 95 to 100)

```python
N = 0
x = 50
for turn in turns:
    x += turn
    if x % 100 == 0:
        N += 1

    r = (x // 100) - ((x-turn)//100)
    N += abs(r)
    if r < 0 and (x-turn) % 100 == 0:
        N -= 1
    if r > 0 and x % 100 == 0:
        N -= 1
N
```

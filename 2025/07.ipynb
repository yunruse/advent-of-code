## Part 1: Beams

Input is comprised of a set of lines. Every even line is just dots and can be ignored. The first line contains an `S` at some index with a "beam". For following lines, if a "beam" hits a "splitter" `^` at index `x` then it is removed and replaced with beams at `x-1` and `x+1`. We can safely assume a splitter is never directly adjacent, but may have a gap of one.

```python
with open("07.txt") as f:
    orig_beam, *splitters = [
        {i for i, c in enumerate(line.strip()) if c != '.'}
        for line in f.readlines()[::2]
    ]
```

```python
beam = set(orig_beam)
N = 0
for split in splitters:
    for s in split:
        if s in beam:
            N += 1
            beam.remove(s)
            beam |= {s-1, s+1}

N
```

# Part 2: Possibilities
Each time a beam is split, we now _count_ the times it is there. Thus if there are two ways to arrive at a destination, we count two paths. In essence, this is the same as in part 1, except that we must now _count_ the paths.

```python
from collections import Counter
N = 1
beam = Counter(orig_beam)
for split in splitters:
    for s in split:
        N += beam[s]
        beam[s-1] += beam[s]
        beam[s+1] += beam[s]
        beam[s] = 0
N
```
